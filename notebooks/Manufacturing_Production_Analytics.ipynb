##Production Analysis

```python
#merge products + production
production_analysis = production.merge(
    products,
    on="product_id",
    how="left"
)
```

```python
#Time variance
production_analysis['time_variance_min'] = (
    production_analysis['actual_batch_time_min'] -
    production_analysis['standard_batch_time_min']
)

production_analysis.head()
```

```python
production_analysis['time_variance_min'].describe()
```

```python
#Rata-rata produk yang memiliki waktu lebih lama dibandingkan standar produksinya
production_analysis.groupby('product_id')['time_variance_min'].mean()
```

```python
#Batch dengan downtime lebih lama memiliki time variance lebih tinggi)
downtime_per_batch = downtime.groupby("batch_id")["downtime_minutes"].sum().reset_index()

downtime_per_batch.head()
```

```python
production_analysis = production_analysis.merge(
    downtime_per_batch,
    on="batch_id",
    how="left"
)

production_analysis.head()
```

```python
production_analysis["downtime_minutes"].isnull().sum()
```

```python
#Ubah NaN jadi 0
production_analysis["downtime_minutes"] = (
    production_analysis["downtime_minutes"].fillna(0)
)
```

```python
production_analysis["downtime_minutes"].isnull().sum()
```

```python
production_analysis["downtime_status"] = production_analysis["downtime_minutes"].apply(
    lambda x: "Has Downtime" if x > 0 else "No Downtime"
)
```

```python
production_analysis.groupby("downtime_status")["time_variance_min"].mean()
```

```python
production_analysis[
    ["downtime_minutes", "time_variance_min"]
].corr()
```
